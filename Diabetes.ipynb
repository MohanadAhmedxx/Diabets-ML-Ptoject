# Classification prediction whether a person is diabetic or non_diabetic

#### Importing imortant modules

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
```

##### getting the csv file

```python
data = pd.read_csv(r"C:\Users\Hp\Desktop\ML and DL\diabetes.csv")
```

```python
print(data.tail())
print("data shape : ", data.shape)
print("#"*100)
data.describe()
```

```python
X = data.drop(columns = "Outcome" , axis = 1)
Y = data["Outcome"]
```

##### sklearn tools for execute ML model

```python
x_train , x_test , y_train, y_test = train_test_split(X,Y,test_size =0.3,stratify=Y,random_state = 5)
print(x_train.shape , x_test.shape)
print(type(x_test))
```

```python
Scale = StandardScaler()
Scale.fit(x_train)
x_train = Scale.transform(x_train)
```

```python
x_train
```

```python
model_1 = SVC(kernel = "linear")
model_1.fit(x_train,y_train)
```

```python
from sklearn.metrics import accuracy_score
```

```python
x_train_prediction = model_1.predict(x_train)
print(accuracy_score(x_train_prediction,y_train))
```

```python
Scale2 = StandardScaler()
x_test_scale = Scale2.fit_transform(x_test)

x_test_prediction = model_1.predict(x_test_scale)
print(accuracy_score(x_test_prediction,y_test))
```

```python
from sklearn.linear_model import LogisticRegression
model_2 = LogisticRegression()
model_2.fit(x_train,y_train)
```

```python
x_train_prediction = model_2.predict(x_train)
print(accuracy_score(x_train_prediction,y_train))
```

```python

x_test_prediction = model_2.predict(x_test_scale)
print(accuracy_score(x_test_prediction,y_test))
```
